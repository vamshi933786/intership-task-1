```python
## 1)
scores = [520, 780, 430]

if scores[0] == scores[1] or scores[1] == scores[2] or scores[0] == scores[2]: ##  code for the tie
    print("Tie Detected")
else:
    highest = max(scores) ## taking the maximum from the scores
    lowest = min(scores)  ## taking the minimum from the scores
    middle = sum(scores) - highest - lowest ## taking the middle (second) from the scores

    print("First Rank :", highest)
    print("Second Rank :", middle)
    print("Third Rank :", lowest)
```

```python
##6)
prices = [250, 180, 450, 120, 600]
total = 0
highest = prices[0]
lowest = prices[0]
for price in prices:
    total = total + price

    if price > highest:
        highest = price

    if price < lowest:
        lowest = price

print("Total Bill:", total)
print("Highest Price:", highest)
print("Lowest Price:", lowest)
```
