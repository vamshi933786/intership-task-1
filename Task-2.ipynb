```python
# Problem 1 –   Student Report Card

students = { 
    "Rahul": 91, 
    "Ayesha": 84, 
    "Kiran": 67, 
    "Ali": 5 
} 

pas = 0
fail = 0
#for loop to iterate through the dictionary and check marks
for student, marks in students.items():
    if marks >=40:
        print(f"{student} has passed with {marks} marks.")
        pas+=1
    else:
        print(f"{student} has failed with {marks} marks.")
        fail+=1
print(f"Total students passed: {pas}")
print(f"Total students failed: {fail}")
```

```python
#   Problem 3 – Login System 
users = { 
    "admin":"python123", 
    "student":"welcome", 
    "guest":"guest123" 
}
T = 3
#checking the username and password
while T>0:
    username = input("Enter your username: ")
    password = input("Enter your password: ")
    if username in users and users[username] == password:
        print(f"Welcome {username}!")
        break
    else:
        T-=1
        print(f"Invalid credentials. You have {T} attempts left.")
```

```python
#Problem 6 – Daily Expense Tracker 
d = {}
while True:
    user = input('Enter Expense Name: ')
    if user.lower() == 'stop':
        break
    amt = int(input('Enter Amount :'))
    d[user] = amt
for k,v in d.items():
    print(k," - ", v)
```

```python
#Problem 9 – Read a Missing File 

try:
    file = open("missing_file.txt", "r")  # Attempt to open a non-existent file
    content = file.read()  # This will raise a FileNotFoundError
    file.close()
except FileNotFoundError:
    content = None

if content:
    print(content)
else:
    print("The file does not exist.")
```

```python
# Problem 10 – Safe Calculator
divisor = int(input("Enter the divisor: "))
dividend = int(input("Enter the dividend: "))
try:
    result = dividend/divisor
    print(f"The result is: {result}")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
#Problem 11 – High Score Board 
scores = { 
    "Rahul":120, 
    "Ali":250, 
    "John":180 
}
print("High Score Board:")
max_score = 0
for player, score in scores.items():
   if score > max_score:
       max_score = score
       top_player = player

print(f"Top Player: {top_player} with a score of {max_score}")
```

```python
# problem 15 – 🛍 Shopping Bill Generator
items = {
    "Milk":50,
    "Bread":40,
    "Eggs":70,
    "Rice":500
}

total_cost = 0

while True:
    item = input("Enter the item name (or type 'done' to finish): ")

    if item.lower() == 'done':
        break

    if item in items:
        total_cost += items[item]

print(f"Total cost of the shopping bill: {total_cost}")
```

```python
# Problem 20 – Space Mission Log (Final Challenge) 
file = open("mission_log.txt", "x")  # Create a new file for mission log
file.close()
```

```python
systems = { 
"Engine":"OK", 
"Fuel":"OK", 
"Navigation":"FAIL", 
"Communication":"OK" 
} 
for k, v in systems.items():
    if v == "FAIL":
        file = open("mission_log.txt", "a")  # Open the file in append mode
        file.write(f"System {k} has failed. Mission cannot proceed.\n")
        file.close()
        print(f"System {k} has failed. Mission cannot proceed.")
        break
    else:
        file = open("mission_log.txt", "a")  # Open the file in append mode
        file.write(f"System {k} is OK. Mission can proceed.\n")
        file.close()
        print(f"System {k} is OK. Mission can proceed.")
```
