```python
## 4)Bonus Calculation
import numpy as np

salary = [25000, 30000, 45000, 28000]

# Python List
updated_salary_list = []
for s in salary:
    updated_salary_list.append(s + 5000)

print("Updated Salary using Python List:")
print(updated_salary_list)

# NumPy Array
salary_arr = np.array(salary)
updated_salary_arr = salary_arr + 5000

print("\nUpdated Salary using NumPy:")
print(updated_salary_arr)
```

```python
## 5)Student Marks Increase
import numpy as np

marks = [60, 75, 80, 90, 65]

# Python List
new_marks = []
for m in marks:
    new_marks.append(m + 5)

print("Python List:")
print(new_marks)

# NumPy Array
marks_arr = np.array(marks)
updated_marks = marks_arr + 5

print("\nNumPy Array:")
print(updated_marks)
```

```python
## 6)Performance Observation
import numpy as np
import time

# Python List
py_list = list(range(100000))

start = time.time()
result_list = [x + 10 for x in py_list]
end = time.time()

print("Python List Time:", end - start)

# NumPy Array
np_array = np.arange(100000)

start = time.time()
result_array = np_array + 10
end = time.time()

print("NumPy Array Time:", end - start)
```

```python
## 19)Arithmetic Operations (UFuncs)
import numpy as np

arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 5, 8])

print("Array 1:", arr1)
print("Array 2:", arr2)

print("Addition:", np.add(arr1, arr2))
print("Subtraction:", np.subtract(arr1, arr2))
print("Multiplication:", np.multiply(arr1, arr2))
print("Division:", np.divide(arr1, arr2))
```

```python
## 21)Trigonometry
import numpy as np

angles = np.array([0, 30, 45, 60, 90])

# Convert degrees to radians
radians = np.radians(angles)

print("Angles:", angles)

print("Sin:", np.sin(radians))
print("Cos:", np.cos(radians))
print("Tan:", np.tan(radians))
```
