```python
## 11)Clean Missing Data
import pandas as pd
import numpy as np

# Create DataFrame
data = {
    "EmployeeID": [101,102,103,104,105,106,107,108,109,110],
    "Name": ["Rahul Sharma","Ayesha Khan","Ali Ahmed","Kiran Kumar","Priya Singh",
             "Rohit Verma","Sneha Reddy","Arjun Rao","Rahul Sharma","Meena Das"],
    "Department": ["HR","IT","Sales","IT","Finance",
                   "HR","IT","Sales","HR","Finance"],
    "Age": [24,27,25,29,26,np.nan,24,31,24,28],
    "Salary": [35000,55000,42000,62000,48000,
               39000,np.nan,65000,35000,51000]
}

df = pd.DataFrame(data)

# Fill missing Age with average Age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill missing Salary with average Salary
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Display updated DataFrame
print(df)
```

```python
## 12)Remove Duplicate Employees
import pandas as pd
import numpy as np

# Create DataFrame
data = {
    "EmployeeID": [101,102,103,104,105,106,107,108,109,110],
    "Name": ["Rahul Sharma","Ayesha Khan","Ali Ahmed","Kiran Kumar","Priya Singh",
             "Rohit Verma","Sneha Reddy","Arjun Rao","Rahul Sharma","Meena Das"],
    "Department": ["HR","IT","Sales","IT","Finance",
                   "HR","IT","Sales","HR","Finance"],
    "Age": [24,27,25,29,26,np.nan,24,31,24,28],
    "Salary": [35000,55000,42000,62000,48000,
               39000,np.nan,65000,35000,51000]
}

df = pd.DataFrame(data)

# Find duplicate records
print("Duplicate Records:")
print(df[df.duplicated()])

# Remove duplicates
df = df.drop_duplicates()

print("\nDataFrame After Removing Duplicates:")
print(df)
```

```python
## 14)oining Year Analysis
import pandas as pd

data = {
    "Name": ["Rahul Sharma", "Ayesha Khan", "Ali Ahmed"],
    "JoiningDate": ["2022-01-15", "2021-08-12", "2023-03-10"]
}

df = pd.DataFrame(data)

# Convert to datetime
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])

# Extract Year and Month
df["JoiningYear"] = df["JoiningDate"].dt.year
df["JoiningMonth"] = df["JoiningDate"].dt.month

print(df)
```
