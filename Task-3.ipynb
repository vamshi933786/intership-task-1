```python
# Problem 4 – Password Inspector 
password = input("Enter a password: ")
c = 0
s = 0
d = 0
for ch in password:
    if ch.isupper():
        c += 1
    elif ch.islower():
        s += 1
    elif ch.isdigit():
        d += 1
if c >= 1 and s >= 1 and d >= 1 and len(password) >= 8:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
#Problem 7 – ATM Withdrawal 
def withdraw(amount, bal):
    if bal < amount:
        print("Insufficient balance")
    else:
        bal-=amount
        print("Withdrawal successful. New balance: ", bal)
withdraw(1000,88000)
```

```python
#Problem 8 – Grade Calculator
def calculate_grade(mark) :
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    else:
        return "F"
print(calculate_grade(49) )
   
```

```python
# Problem 9 – 🛍 Bill Calculator 
def calculate_bill(price, quantity):
    return price * quantity
print(calculate_bill(10, 5) )

```

```python
# Problem 10 – Discount Calculator 
discount = lambda p, d: p * (1 - d / 100)
print(discount(1500, 10))  # Example usage: calculate 10% discount on $100
```

```python
# Problem 11 –  Rectangle Area 
rec = lambda l, w: l * w
print(rec(5, 10))  # Example usage: calculate area of rectangle with length 5 and width 10
```
