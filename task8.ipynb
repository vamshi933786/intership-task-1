```python
## 1)Student Marks Analysis
import pandas as pd

# Python List
marks_list = [78, 85, 91, 67, 88]
print("Python List:")
print(marks_list)

# Pandas Series
marks_series = pd.Series(marks_list)
print("\nPandas Series:")
print(marks_series)

# Advantages of Pandas
print("\nAdvantages of Pandas:")
print("1. Easy data analysis.")
print("2. Fast filtering and searching.")
print("3. Supports reading CSV, Excel, and JSON files.")
```

```python
## 8)Temperature Report
import pandas as pd

temperature = pd.Series([28, 31, 29, 35, 33, 27, 30])

print("Temperatures:")
print(temperature)

print("\nTemperatures Greater Than 30°C:")
print(temperature[temperature > 30])
```

```python
## 17)Dataset Information
import pandas as pd

# Sample dataset
data = {
    "Name": ["Rahul", "Ayesha", "Ali", "Kiran", "John"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)

print("Column Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nTotal Rows:")
print(len(df))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
df.info()
```
